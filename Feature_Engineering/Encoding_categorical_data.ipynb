*-Numerical data*  
*-Categorical data*  

*categorical data can be further divided into:*  
*- Nominal data*   
*- categorical data*  

**Encoding is the process of converting categorical data into numerical as most of the ml algos expect data in numerical forms**

*for ordinal data ---> Ordinal encoding/(label encoding)*  
*for nominal data ---> one hot encoding*  

Ordinal encoder vs label encoder

*for input variable we use ordinal encoder*  
*for target variable (output) we use label encoder*

## One Hot Encoding

*Why can't we encode nominal data as we do ordinal data?*

*Because our machine learning models may interpret nominal data as having an order. For instance, if there are three colors—let's say Red, Green, and Blue—and we encode them as Red = 0, Green = 1, and Blue = 2, the model might assume that Blue is greater or more important than Red and Green, which is not the case.*

![one hot encoding](ohe.png)

***Dummy Variable Trap***

- *After doing one hot encoding, we have to remove one of the dummy columns to remove the problem of multicolinearity*
- *to make all columns independent*

```python
data = pd.read_csv('cars.csv')
data
```

```python
type(data)
```

```python
data.brand.value_counts()
```

```python
data.fuel.value_counts()
```

*we can see that owner and fuel has less nominal value where brand has many, so if we apply OHE on brand then many new cols about 40 as different brands are created which makes very high no of columns so processing becomes hard and slow in such case we have to group uncommon i.e less frequent columns into single to reduce dimensonality*

## **1. One Hot Encoding using Pandas**

```python
pd.get_dummies(data,columns=['fuel','owner'],dtype=int) #not doing for brand now as it contains many values
```

## **2. K-1 OneHotEncoding**

*still there is problem of multicolinearity*

```python
pd.get_dummies(data,columns=['fuel','owner'],dtype=int,drop_first=True)
```

## **4. OneHotEncoding with Top Categories**

```python
counts = data['brand'].value_counts()
counts
```

```python
data['brand'].nunique()
threshold = 100
```

```python
replace = counts[counts <= threshold].index
replace
```

```python
pd.get_dummies(data['brand'].replace(replace, 'uncommon'),dtype=int).sample(5)
```
