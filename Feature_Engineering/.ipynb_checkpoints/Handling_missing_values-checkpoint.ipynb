# **Handling Missing Values**

*Handling missing value is very important in ml and data science , as must of the algorithm cannot handle missing values*

*There are two ways of handling it*
- **Removing all rows containing missing values**
- **Imputing values**

## **Removing all rows containing missing values**

*This is the most simple way to handle missing value we just remove all the rows which contains missing values. It has many disadvantages so it is not prefred this technique to handle data efficeintly.*

**Thinks to take care while removing missing values rows**
- Missing value should be completely random
- the total percentage of missing data should be low as possible

## CCA

```python
#lets import data againa
df = pd.read_csv('data_science_job.csv')
df.head()
```

```python
df.isnull().mean()*100
```

*we can see that columns like [company_type ,company_size,major_discipline,gender] has high number of missing values so we cannot remove the rows completeley*

*we will remove values where null values is less than 5%*

```python
cols = [i for i in df.columns if df[i].isnull().mean()<0.05 and df[i].isnull().mean()>0]
cols
```

```python
df[cols].sample(5)
```

```python
len(df[cols].dropna()) / len(df) #ratio of df after removing missing values in cols to the original
```

*still our 89% data will be preserved , we will lost only 11%*

```python
new_df = df[cols].dropna()
```

```python
df.shape,new_df.shape
```

```python
19158 - 17182
```

1976 rows are removed 

```python
df['training_hours'].hist()
```

```python
new_df['training_hours'].hist(color = 'red')
```

training_hours columns has almost same distribution after removing the missing values

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(nrows=5, ncols=2, figsize=(12, 20), squeeze=False)

for i, col in enumerate(cols):
    # Left subplot (df)
    if df[col].dtype == 'object':  # if column is categorical
        df[col].value_counts().plot(kind='bar', ax=axes[i, 0], color='blue', alpha=0.6)
    else:  # if column is numeric
        axes[i, 0].hist(df[col].dropna(), color='blue', alpha=0.6)
    axes[i, 0].set_title(f"Histogram of {col} (df)")
    
    # Right subplot (new_df)
    if new_df[col].dtype == 'object':
        new_df[col].value_counts().plot(kind='bar', ax=axes[i, 1], color='red', alpha=0.6)
    else:
        axes[i, 1].hist(new_df[col].dropna(), color='red', alpha=0.6)
    axes[i, 1].set_title(f"Histogram of {col} (new_df)")

plt.tight_layout()
plt.show()
```

*after removing all the missing values data distribution is alike, almost similar, so we can do CCA in this case*

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

fig, axes = plt.subplots(nrows=5, ncols=2, figsize=(12, 20), squeeze=False)

for i, col in enumerate(cols):
    # Left subplot (df)
    if df[col].dtype == 'object':  # Categorical
        df[col].value_counts(normalize=True).plot(kind='bar', ax=axes[i, 0], color='blue', alpha=0.6)
    else:  # Numeric
        sns.kdeplot(df[col].dropna(), ax=axes[i, 0], color='blue', fill=True, alpha=0.3)
    axes[i, 0].set_title(f"PDF of {col} (df)")
    
    # Right subplot (new_df)
    if new_df[col].dtype == 'object':
        new_df[col].value_counts(normalize=True).plot(kind='bar', ax=axes[i, 1], color='red', alpha=0.6)
    else:
        sns.kdeplot(new_df[col].dropna(), ax=axes[i, 1], color='red', fill=True, alpha=0.3)
    axes[i, 1].set_title(f"PDF of {col} (new_df)")

plt.tight_layout()
plt.show()
```

## **Imputing missing Values**

*another technique for handling missing values is imputing which is much better than CCA,*

### Imputing for Numerical column

#### **Univariate Imputation**

*If we used the values from the same column to fill the missing value then it is Univariate Imputation*

***Example:*** : *if we use mean/median or any statistical technique to fill the missing value of the same column*

#### **Multivariate Imputation**

*If we used the values from the other columns and same columns to fill the missing values then it is Multivariate Imputation*

### **Univariate Imputation for Numerical data**

*Techniques*
- mean/median
- arbitary value
- random
- end of distribution etc

#### Mean/median Imputation

- use mean for normal distribution , i.e mean and median is same for normal distribution
- use median for skwed data i.e data with outliers

**Benifts of Mean/median Imputaion**
- simple
- simple on production too

**Disadvatage of mean/median imputation**
- Change the distribution of data
- extra outliers were intoduced
- changes in correlation/covariance

**When to use Mean/Median Imputation**
- MCAR(Missing completely at random just like CCA)
- small amount of data (5%) is missing

### Code Example : Using Pandas

```python
df = pd.read_csv('titanic_toy.csv')
df.head()
```

```python
df.isnull().sum()
```

*Age and Fare has some missing values*

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
X_train.shape,X_test.shape
```

```python
y_train.shape,y_test.shape
```

```python
X_train.isnull().mean()
```

*Age column has 20% of missing values and Fare has 5%*

```python
# calculating mean and median value for both column
age_median= X_train['Age'].median()
age_mean = X_train['Age'].mean()

fare_median = X_train['Fare'].median()
fare_mean = X_train['Fare'].mean()
```

```python
X_train['Age_median'] = X_train['Age'].fillna(age_median)
X_train['Age_mean'] = X_train['Age'].fillna(age_mean)

X_train['Fare_median'] = X_train['Fare'].fillna(fare_median)
X_train['Fare_mean'] = X_train['Fare'].fillna(fare_mean)
```

```python
X_train.sample(5)
```

*we have filled all the missing values with mean and median for age and fare and created new columns as above*

```python
# now lets calculate the variance of original data and imputed data
print('Original Age variable variance: ', X_train['Age'].var())
print('Age Variance after median imputation: ', X_train['Age_median'].var())
print('Age Variance after mean imputation: ', X_train['Age_mean'].var())

print('Original Fare variable variance: ', X_train['Fare'].var())
print('Fare Variance after median imputation: ', X_train['Fare_median'].var())
print('Fare Variance after mean imputation: ', X_train['Fare_mean'].var())
```

**Note:** *there is drastic changes in the age variance for original data and after median and mean imputaition, it is important to check the variance after imputation so that we can see the changes in our data`s distribution, in this case age imputation using mean and median is not quite good as it has drastic difference in varaince*

*while Fare has almost similar variance because there are few missing values in fare column it is good to do mean and median imputation in fare column*

```python
fig = plt.figure()
ax = fig.add_subplot(111)

# original variable distribution
X_train['Age'].plot(kind='kde', ax=ax)

# variable imputed with the median
X_train['Age_median'].plot(kind='kde', ax=ax, color='red')

# variable imputed with the mean
X_train['Age_mean'].plot(kind='kde', ax=ax, color='green')

# add legends
lines, labels = ax.get_legend_handles_labels()
ax.legend(lines, labels, loc='best')
```

*From this figure also we can clearly see that distriution of original data is different than other two data*

```python
fig = plt.figure()
ax = fig.add_subplot(111)

# original variable distribution
X_train['Fare'].plot(kind='kde', ax=ax)

# variable imputed with the median
X_train['Fare_median'].plot(kind='kde', ax=ax, color='red')

# variable imputed with the mean
X_train['Fare_mean'].plot(kind='kde', ax=ax, color='green')

# add legends
lines, labels = ax.get_legend_handles_labels()
ax.legend(lines, labels, loc='best')
```

*In this figure we can see almost distribution is same i.e there is no impact of imputation on the original data distributions which is very good for us*

```python
# lets see the covariance
X_train.cov()
```

*we can see that covariance of age_mean/age_median to other column is different than original age column which is the red flag*

```python
#correlation
X_train.corr()
```

```python
X_train[['Age', 'Age_median', 'Age_mean']].boxplot()
```

*Age_median/Age_mean has introduced alot of outliers than original Age which is also a subject of concern*

```python

X_train[['Fare', 'Fare_median', 'Fare_mean']].boxplot()
```

*whereas Fare_median/Fare_mean looks almost same there is no any drastic changes as compared to original ones*

***Now we can conclude that if we have less missing values less than 5-6% then our mean/median imputation just works perfect but if we apply it on large data then it might change our data`s distribution and we may lost essense of our data***

### Code Example Using Scikit Learn

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer1 = SimpleImputer(strategy='median')
imputer2 = SimpleImputer(strategy='mean')
```

```python
from sklearn.compose import ColumnTransformer
trf = ColumnTransformer([
    ('imputer1',imputer1,['Age']),
    ('imputer2',imputer2,['Fare'])
],remainder='passthrough')
```

```python
trf.fit(X_train)
```

```python
trf.named_transformers_['imputer1'].statistics_
```

```python
trf.named_transformers_['imputer2'].statistics_
```

```python
X_train = trf.transform(X_train)
X_test = trf.transform(X_test)
```

```python
X_train
```

*it is better to use sklearn instead of python because we can create a pipelines and in production also if we get new data it can be prerocessed easily*

### **Handling Categorical Missing Data**

Categorical missing values can be replaced in two ways
- By most frequent item (mode)
- By creating new category (missing)

### By most frequent item (mode)

*make sure to use this technique the data much is mode should be highly frequent than other*

## **Random Imputation**

*in this technique we fill missing value by taking random numbers from the same column,i.e we will take any arbitary value from the same column*

*it can be used to fill numerical and categorical missing values*

**Benifits of Random Imputation**

- easy
- Variance and distribution of data is not changed after imputation , it would be almost similar to the original data
- well suited for linear models

**Disadvatage of Random Imputaion**

- memory heavy for deployment , as we have to store all the training data in the server for filling missing value in test (unseen) data.
