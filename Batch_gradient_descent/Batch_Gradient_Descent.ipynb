# **Types of Gradient Descent**

*Generally, types of gradient descent can be divided in three types*
- Batch Gradient Descent
- stochastic Gradient descent
- mini batch gradient descent  

**Batch Gradient Descent** : *A type of gradient descent where the entire dataset (all rows) is used to compute the gradient and update the coefficients once per iteration.*

**Stochastic Gradient Descent** : *A type of gradient descent where one data point (a single row) is used to compute the gradient and update the coefficients. This happens once per row, so multiple updates per epoch*

**Mini Batch Gradient Descent** : *A type of gradient descent where a small batch of rows (e.g., 30 rows at a time) is used to compute the gradient and update the coefficients.*
- **Example**: *If you have 300 rows and batch size = 30, you’ll have 10 updates per epoch.*

# **Batch Gradient Descent**

**Pros**:
- ✅ Stable convergence

**Cons**:
- ❌ Can be very slow for large datasets
- ❌ Memory-intensive

### 📘 Batch Gradient Descent in Multiple Linear Regression

Suppose we are solving a **Multiple Linear Regression** problem with an **n-dimensional input vector**.

The hypothesis (prediction) function is:

$$
\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_n x_n
$$

Or in **vectorized form**:

$$
\hat{y} = \boldsymbol{\beta}^\top \mathbf{x}
$$

Where:
- $\boldsymbol{\beta} = [\beta_0, \beta_1, \ldots, \beta_n]^\top$ is the **(n+1)-dimensional coefficient vector**
- $\mathbf{x} = [1, x_1, x_2, \ldots, x_n]^\top$ includes the **bias term** (1 for $\beta_0$)

---

### 🎯 Goal
Minimize the **Mean Squared Error (MSE)** cost function:

$$
J(\boldsymbol{\beta}) = \frac{1}{2m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right)^2
$$

Where:
- $m$ = number of training examples
- $\hat{y}^{(i)} = \boldsymbol{\beta}^\top \mathbf{x}^{(i)}$

---

### 🧮 Partial Derivatives of the Cost Function

To perform gradient descent, we compute the **partial derivative** of the cost function with respect to each parameter $\beta_j$:

$$
\frac{\partial J}{\partial \beta_j} = \frac{1}{m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right) x_j^{(i)}
$$

This gives the direction in which the parameter $\beta_j$ should be adjusted to minimize the cost.

---

### 🔁 Batch Gradient Descent Algorithm

At each iteration, update each coefficient using the partial derivative:

$$
\beta_j := \beta_j - \alpha \cdot \frac{\partial J}{\partial \beta_j}
= \beta_j - \alpha \cdot \frac{1}{m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right) x_j^{(i)}
\quad \text{for } j = 0, 1, \ldots, n
$$

Where:
- $\alpha$ = learning rate
- $x_0^{(i)} = 1$ (for the intercept $\beta_0$)

---

### ✅ Example: 3-Dimensional Input

Let’s consider:

$$
\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 x_3
$$

Then the partial derivatives are:

$$
\frac{\partial J}{\partial \beta_0} = \frac{1}{m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right)
$$

$$
\frac{\partial J}{\partial \beta_1} = \frac{1}{m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right) x_1^{(i)}
$$

$$
\frac{\partial J}{\partial \beta_2} = \frac{1}{m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right) x_2^{(i)}
$$

$$
\frac{\partial J}{\partial \beta_3} = \frac{1}{m} \sum_{i=1}^{m} \left( \hat{y}^{(i)} - y^{(i)} \right) x_3^{(i)}
$$

So the update rules become:

$$
\beta_0 := \beta_0 - \alpha \cdot \frac{\partial J}{\partial \beta_0}
$$

$$
\beta_1 := \beta_1 - \alpha \cdot \frac{\partial J}{\partial \beta_1}
$$

$$
\beta_2 := \beta_2 - \alpha \cdot \frac{\partial J}{\partial \beta_2}
$$

$$
\beta_3 := \beta_3 - \alpha \cdot \frac{\partial J}{\partial \beta_3}
$$

---

### 📌 Key Characteristics of Batch Gradient Descent

**Pros**:
- ✅ Stable convergence
- ✅ Accurate gradient calculation (based on full dataset)

**Cons**:
- ❌ Slow for large datasets
- ❌ Requires loading the entire dataset into memory

---

### 🔍 Summary

Batch Gradient Descent updates all coefficients **after computing the gradients from the full dataset**.  
Each update is based on the **partial derivative of the cost function with respect to each coefficient**, ensuring the model moves in the direction of steepest descent.

It is stable and accurate, but computationally expensive for large datasets.

**Using own implementation form scratch**

```python
class GDRegressor:
    
    def __init__(self,learning_rate=0.01,epochs=100):
        
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        
    def fit(self,X_train,y_train):
        # init your coefs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            # update all the coef and the intercept
            y_hat = np.dot(X_train,self.coef_) + self.intercept_
            #print("Shape of y_hat",y_hat.shape)
            intercept_der = -2 * np.mean(y_train - y_hat)
            self.intercept_ = self.intercept_ - (self.lr * intercept_der)
            
            coef_der = -2 * np.dot((y_train - y_hat),X_train)/X_train.shape[0]
            self.coef_ = self.coef_ - (self.lr * coef_der)
        
        print(self.intercept_,self.coef_)
    
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
gdr = GDRegressor(epochs=1000,learning_rate=0.5)
```

```python
gdr.fit(X_train,y_train)
```

```python
y_pred = gdr.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
